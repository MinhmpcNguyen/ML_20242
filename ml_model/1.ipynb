```python
import zipfile
import os

zip_path = "dogs-vs-cats-redux-kernels-edition.zip"
extract_to = "dogs-vs-cats"  # tên thư mục sau khi giải nén

# Giải nén zip
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_to)

print("Đã giải nén xong vào thư mục:", extract_to)
```

```python
# Kiểm tra folder sau khi giải nén
for root, dirs, files in os.walk("dogs-vs-cats"):
    print("📂", root)
    for file in files[:5]:  # chỉ in 5 file đầu tiên trong mỗi folder
        print("   └──", file)
```

```python
import zipfile
import os

# Tạo thư mục đích nếu chưa có
os.makedirs("dogs-vs-cats/train", exist_ok=True)
os.makedirs("dogs-vs-cats/test", exist_ok=True)

# Giải nén train.zip
with zipfile.ZipFile("dogs-vs-cats/train.zip", 'r') as zip_ref:
    zip_ref.extractall("dogs-vs-cats/train")

# Giải nén test.zip
with zipfile.ZipFile("dogs-vs-cats/test.zip", 'r') as zip_ref:
    zip_ref.extractall("dogs-vs-cats/test")

print("✔️ Đã giải nén xong!")
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import glob

# Đọc ảnh từ thư mục train
image_paths = glob.glob("dogs-vs-cats/train/*.jpg")[:6]  # chỉ lấy 6 ảnh đầu

# Hiển thị ảnh theo lưới 2 hàng x 3 cột
plt.figure(figsize=(12, 6))
for i, path in enumerate(image_paths):
    img = Image.open(path)
    plt.subplot(2, 3, i + 1)
    plt.imshow(img)
    plt.title(os.path.basename(path))
    plt.axis("off")
plt.tight_layout()
plt.show()
```
